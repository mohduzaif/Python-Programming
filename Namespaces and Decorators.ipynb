### Namespaces

A namespace is a space that holds names(identifiers).Programmatically speaking, namespaces are dictionary of identifiers(keys) and their objects(values)

There are 4 types of namespaces:
- Builtin Namespace
- Global Namespace
- Enclosing Namespace
- Local Namespace

### Scope and LEGB Rule

A scope is a textual region of a Python program where a namespace is directly accessible.

The interpreter searches for a name from the inside out, looking in the local, enclosing, global, and finally the built-in scope. If the interpreter doesn’t find the name in any of these locations, then Python raises a NameError exception.

```python
# local and global Variables. 

# global variable. 
a = 1

def function(): 
    # local Variable.
    b = 2
    print('From Function : ', b)

function()
print('From Global Area : ', a) 
```

```python
# Enclosing scope

# outer function is a Enclosing scope. 
# inner function is a Local Scope. 
# And all the things outside the outer is a global scope. 

def outer():
  def inner():
    print(a)
  inner()
  print('outer function')


outer()
print('main program')
```

```python
# nonlocal keyword
def outer():
  a = 1
  def inner():
    nonlocal a
    a += 1
    print('inner',a)
  inner()
  print('outer',a)


outer()
print('main program')
```

### Decorators

A decorator in python is a function that receives another function as input and adds some functionality(decoration) to and it and returns it.

This can happen only because python functions are 1st class citizens (we do all the operation that we are did with the datatypes of python).

There are 2 types of decorators available in python
- `Built in decorators` like `@staticmethod`, `@classmethod`, `@abstractmethod` and `@property` etc
- `User defined decorators` that we programmers can create according to our needs

```python
# Python are 1st class function

def modify(func,num):
  return func(num)

def square(num):
  return num**2

modify(square,2)
```

```python
def my_decorator(func): 
    def wrapper(): 
        print('********************************')
        func() 
        print('********************************') 
        
    return wrapper

def temp(): 
    print('Hello world')

def print_number(): 
    for i in range(1, 11): 
        print(i, end=' ')
    print()

t = my_decorator(temp)
t()

pn = my_decorator(print_number)
pn()
```

```python
# Better syntax for the above example
# simple example 

def my_decorator(func): 
    def wrapper(): 
        print('********************************')
        func() 
        print('********************************') 
        
    return wrapper

# just type this above the function defination.
@my_decorator
def temp(): 
    print('Hello world')

# just type this above the function defination.
@my_decorator
def print_number(): 
    for i in range(1, 11): 
        print(i, end=' ')
    print()

temp()
```

```python
print_number()
```

```python
# Create a decorator which calculate the time of any function. 

import time 

def timer_decorator(func): 
    def wrapper(): 
        start = time.time() 
        func()
        end = time.time() 
    
        print("Time taken by the function'", func.__name__, "'is", end - start, "seconds")

    return wrapper

@timer_decorator
def display_name(): 
    print('Hello Uzaif')
    time.sleep(3)

@timer_decorator
def greet(): 
    print('Good Morning!!')
    time.sleep(2)


display_name()

greet()
```

`NOTE : ` The problem in the above code it will execute for those function which does not have any aruguments. 

```python
# Create a decorator which calculate the time of any function with having arguments. 
import time 

def timer_decorator(func): 
    def wrapper(*args): 
        start = time.time() 
        func(*args)
        end = time.time() 
    
        print("Time taken by the function'", func.__name__, "'is", end - start, "seconds")

    return wrapper

@timer_decorator
def greet(name): 
    print('Good Morning,', name)
    time.sleep(2)

@timer_decorator
def square(num): 
    time.sleep(3)
    print(num ** 2)

@timer_decorator
def power(num1, num2): 
    time.sleep(1)
    print(num1 ** num2)


greet('Huzaifa')
square(3)
power(2, 3)
```

```python
# A big problem
# One last example -> decorators with arguments

def sanity_check(data_type):
  def outer_wrapper(func):
    def inner_wrapper(*args):
      if type(*args) == data_type:
        func(*args)
      else:
        raise TypeError('Datatype of the input data is not correct.')
    return inner_wrapper
  return outer_wrapper

@sanity_check(int)
def square(num):
  print(num**2)

@sanity_check(str)
def greet(name):
  print('Hello',name)

square(2)
greet('Uzaif')
```

## How this works -> closure?

### Structure of a Closure
- To understand closures, you need to understand three things:

- Nested functions – functions defined inside another function.

- Free variables – variables that are used in the inner function but defined in the outer function.

- Returning the inner function – the outer function must return the inner function.

### Closure Conditions
- A closure exists when:

 - A nested function is defined inside another function.

 - The inner function uses variables from the outer function.

 - The outer function returns the inner function.

Example below : 

```python
def counter():
    
    count = 0
    def increment(): 
        nonlocal count
        count += 1
        
        return count

    return increment

c = counter()

print(c())
print(c())
print(c())
print(c())
```
